```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 1, 100)

def my_sin(x):
    return np.sin(x)

def my_cos(x):
    return np.cos(x)
```

```python
y_sin = my_sin(x)
y_cos = my_cos(x)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
ax1.plot(x, y_sin, color='orange')
ax1.set_title('sin(x)')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)')
ax2.plot(x, y_cos, color='blue')
ax2.set_title('cos(x)')
ax2.set_xlabel('x')
ax2.set_ylabel('cos(x)')
plt.tight_layout()
plt.savefig("sin_cos_plot.pdf")
plt.show()
```
